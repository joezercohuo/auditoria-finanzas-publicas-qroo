```python
import pandas as pd

# 1. Cargar el archivo a la memoria de Python
df_contratos = pd.read_csv('detalle_formato_37_parte1.csv', encoding='latin1')

# 2. Forzar a pandas a mostrar todas las columnas sin ocultarlas
pd.set_option('display.max_columns', None)

# 3. Diagonostico 1: Tamaño exacto de la base
print("Total de filas (Filas, Columnas):", df_contratos.shape)

# 4. Diagnostico 2: Tipos de datos y valores vacíos (radiografia)
print("\n--- INFORMACIÓN DE COLUMNAS ---")
df_contratos.info()

# 5. Diagnostico 3: Previsualización de los datos reales
display(df_contratos.head(3))
```

```python
# 1. Recargamos el archivo con la codificación exacta para arreglar los acentos mexicanos
# El parámetro 'utf-8-sig' es el que elimina el símbolo ï»¿ y repara las tildes
df_contratos = pd.read_csv('detalle_formato_37_parte1.csv', encoding='utf-8-sig')

# 2. Vamos a extraer e imprimir la lista de las 95 columnas numeradas
# Usamos un bucle 'for' para leerlas hacia abajo y facilitar la búsqueda visual
print("--- LISTA COMPLETA DE VARIABLES ---")
columnas_totales = df_contratos.columns.tolist()

for indice, nombre_columna in enumerate(columnas_totales):
    print(f"[{indice}] {nombre_columna}")
```

```python
# 1. Filtramos la tabla original pasándole una lista con los nombres exactos de las 4 columnas
columnas_utiles = [
    'Tipo de procedimiento (catálogo)',
    'Denominación o razón social',
    'Fecha del contrato expresada con el formato día/mes/año',
    'Monto total del contrato con impuestos incluidos (MXN)'
]

# Creamos un nuevo DataFrame (tabla) llamado df_limpio solo con esos datos
df_limpio = df_contratos[columnas_utiles].copy()

# 2. Renombramos los encabezados burocráticos a nombres técnicos cortos usando un diccionario {}
df_limpio = df_limpio.rename(columns={
    'Tipo de procedimiento (catálogo)': 'tipo_procedimiento',
    'Denominación o razón social': 'proveedor',
    'Fecha del contrato expresada con el formato día/mes/año': 'fecha',
    'Monto total del contrato con impuestos incluidos (MXN)': 'monto_total'
})

# 3. Previsualizamos nuestra nueva base de datos optimizada
display(df_limpio.head(5))
```

```python
# 1. Convertir la columna 'fecha' a un formato de tiempo oficial (Datetime)
# errors='coerce' fuerza a que si hay un texto mal escrito en lugar de fecha, lo deje en blanco en vez de colapsar el programa
df_limpio['fecha'] = pd.to_datetime(df_limpio['fecha'], errors='coerce')

# 2. Asegurar que los montos sean estrictamente valores numéricos (decimales)
df_limpio['monto_total'] = pd.to_numeric(df_limpio['monto_total'], errors='coerce')

# 3. Revisar si quedó algún valor nulo (celdas vacías) después de la conversión
print("--- DATOS FALTANTES ---")
print(df_limpio.isnull().sum())

# 4. Exportar nuestra base de datos limpia a un nuevo archivo CSV
# index=False evita que se guarde la primera columna de números (0, 1, 2, 3...) que Python genera automáticamente
df_limpio.to_csv('contratos_limpios_qroo.csv', index=False, encoding='utf-8-sig')

print("\n¡Archivo exportado con éxito para Power BI!")
```

```python
from scipy import stats

# 1. Separar los montos de contratos en dos grupos según el procedimiento
adjudicaciones = df_limpio[df_limpio['tipo_procedimiento'] == 'Adjudicación directa']['monto_total'].dropna()
licitaciones = df_limpio[df_limpio['tipo_procedimiento'] == 'Licitación pública']['monto_total'].dropna()

# 2. Calcular los promedios monetarios de cada grupo
print("--- PROMEDIOS POR TIPO DE CONTRATO ---")
print(f"Promedio Adjudicación Directa: ${adjudicaciones.mean():,.2f}")
print(f"Promedio Licitación Pública:   ${licitaciones.mean():,.2f}")

# 3. Aplicar T-test de Welch (compara las medias de dos grupos independientes)
t_stat, p_value = stats.ttest_ind(adjudicaciones, licitaciones, equal_var=False)

print("\n--- PRUEBA ESTADÍSTICA (T-TEST) ---")
print(f"Estadístico T: {t_stat:.4f}")
print(f"Valor p (p-value): {p_value:.5f}")
```
